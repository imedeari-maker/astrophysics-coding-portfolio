```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

stellarModel = '1msol-z/hr.0A'

dataframe_hr = pd.read_table(stellarModel, sep=r'\s+', names=np.arange(14))
nrows = np.shape(dataframe_hr)[0]

T_array_hr = []
L_array_hr = []
t_array_hr = []
r_array_hr = []
for i in range(nrows):
    if i%19 == 0:
        T_array_hr.append(dataframe_hr[8][i])
        L_array_hr.append(dataframe_hr[7][i])
        t_array_hr.append(dataframe_hr[2][i])
        r_array_hr.append(dataframe_hr[4][i])

plt.xlim()
plt.plot(T_array_hr, L_array_hr, color='blue')
plt.title('Hertzsprung-Russell Diagram')
plt.xlabel('log(Temperature) [T]')
plt.ylabel('log(Luminosity) [L]')
plt.gca().invert_xaxis()
plt.show()


```
