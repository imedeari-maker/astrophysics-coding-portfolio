```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Set up grid to match paper
x_min, x_max = 0, 26
y_min, y_max = 0, 23
grid_resolution = 300 

x = np.linspace(x_min, x_max, grid_resolution)
y = np.linspace(y_min, y_max, grid_resolution)
X, Y =np.meshgrid(x, y)

# 2. Define Positions
x_center, y_center = 14.0, 10.0 # Center of circle
y_bar = 1 # Vertical position of bottom bar

# 3. Physics simulation using mirror method
r1_sq = (X - x_center)**2 + (Y - y_center)**2 +1e-6
r2_sq = (X - x_center)**2 + (Y - (2 * y_bar - y_center))**2 + 1e-6

# Calculate x and y componenets
Ex = -(X - x_center)/r1_sq - (X - x_center)/r2_sq
Ey = -(Y - y_center)/r1_sq - (Y - (2 * y_bar - y_center))/r2_sq

# 4. Mask out inside of circle and bar
inside_circle = (X - x_center)**2 + (Y - y_center)**2 < 1.0
inside_bar = (Y <= 1.2) & (X >= 5) & (X <= 23)
Ex[inside_circle | inside_bar] = np.nan
Ey[inside_circle | inside_bar] = np.nan

# 5. Create clean lines
fig, ax = plt.subplots(figsize=(9, 8.5), dpi=300) 
ax.set_facecolor('#ffffff')

# Draw physical shapes
circle = plt.Circle((x_center, y_center), 0.8, color='#2c3e50', zorder=5, label='Circular Electrode (+)')
ax.add_patch(circle)
ax.plot([5, 23], [y_bar, y_bar], color='#2c3e50', linewidth=10, solid_capstyle='butt', zorder=5, label='Bar Electrode (-)')

# Draw the electric field lines
ax.streamplot(X, Y, Ex, Ey,
              color='#e74c3c',
              linewidth=1.2,
              density=1.3,
              arrowstyle='->',
              arrowsize=1.2)

# 6. Formatting to match lab report
ax.set_xlim(x_min, x_max)
ax.set_ylim(y_min, y_max)
ax.set_xticks(np.arange(0, 27, 2))
ax.set_yticks(np.arange(0, 21, 2))
ax.set_xlabel('X-Coordinate (cm)', fontsize=11, fontweight='bold')
ax.set_ylabel('Y-Coordinate (cm)', fontsize=11, fontweight='bold')
ax.set_title('Experimentally Derived Electric Field Mapping\n(Point-to-Plane Configuration)', fontsize=13, fontweight='bold', pad=15)
ax.grid(True, which='both', color='#e0e0e0', linestyle='--', linewidth=0.7)
ax.set_aspect('equal')

# For saving file
plt.savefig('electric_field_plot.png', bbox_inches='tight', dpi=300)
plt.show()
```
