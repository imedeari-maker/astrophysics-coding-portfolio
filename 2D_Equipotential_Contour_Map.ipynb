```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import griddata

# 1. Complete 30 experimental data points from Part A
data = np.array([
[14, 8, -0.832], [14, 5.5, 0.300], [18, 5, -0.292], [18, 10, -0.159],
[18, 14, -0.113], [10, 14, -0.108], [8, 12, -0.099], [6, 10, -0.127],
[6, 6, -0.180], [14, 13, -0.237], [25, 5, -0.007], [25, 10, -0.045],
[25, 15, -0.014], [20, 18, -0.022], [17, 18, -0.037], [26, 18, -0.005],
[2, 18, -0.003], [6, 12, -0.053], [5, 3, -0.492], [2, 4, -0.192],
[15, 3, -0.302], [13, 3, -0.337], [5, 16, -0.017], [1, 19, -0.001],
[27, 10, -0.002], [27, 10, -0.001], [1, 1, 0.000], [15, 9, -0.850],
[10, 8, -0.300], [13, 11, -0.547]
])

x_measured, y_measured, v_measured = data[:, 0], data[:, 1], data[:, 2]

# 2. Recreate 26x20 lab grid
x_grid = np.linspace(0, 26, 200)
y_grid = np.linspace(0, 20, 200)
X, Y = np.meshgrid(x_grid, y_grid)

# 3. Linear interpolation to fill the gaps
V = griddata((x_measured, y_measured), v_measured, (X, Y), method='linear')
V_nearest = griddata((x_measured, y_measured), v_measured, (X, Y), method='nearest')
V[np.isnan(V)] = V_nearest[np.isnan(V)]

# 4. Plotting the Part A 2D Equipotential Map
fig, ax = plt.subplots(figsize=(10, 8), dpi=300)
ax.set_facecolor('#ffffff')

# Draw the 30 physical points as dots to show data locations
ax.scatter(x_measured, y_measured, color='blue', s=15, zorder=5, label='Measured Points')

# Set contour intervals every 0.1V to capture decimal ranges neatly
levels = np.arange(-0.8, 0.4, 0.1)
contours = ax.contour(X, Y, V, levels=levels, cmap='plasma', linewidths=1.5, zorder=3)

# Label execution to prevent library version crashes
ax.clabel(contours, inline=True, fmt='%.2fV', colors='black')

# Grid and axis formatting to match canvas sheet
ax.set_xlim(0, 26)
ax.set_ylim(0, 20)
ax.set_xticks(np.arange(0, 27, 2))
ax.set_yticks(np.arange(0, 21, 2))
ax.set_xlabel('X-Coordinate (cm)', fontsize=11, fontweight='bold')
ax.set_ylabel('Y-Coordinate (cm)', fontsize=11, fontweight='bold')
ax.set_title('2-D Equipotential Contour Map (Part A - 30 Points)\nDerived from Real Laboratory Measurements', fontsize=12, fontweight='bold', pad=15)
ax.grid(True, which='both', color='#e0e0e0', linestyle='--', linewidth=0.5)
ax.set_aspect('equal')

# Save the high-res 2D map image
plt.savefig('2d_partA_contours.png', bbox_inches='tight', dpi=300)
plt.show()
```
