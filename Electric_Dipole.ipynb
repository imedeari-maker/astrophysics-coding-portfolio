```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. Constants & Scale ---
kcoulomb = 9e9
qproton = 1.6e-19
scale = 1e10 # Convert meters to Angstroms

# --- 2. Create the Grid (No loops!) ---
# This builds a clean grid from -3 to 3 Angstroms
x = np.linspace(-3, 3, 5)
y = np.linspace(-3, 3, 5)
z = np.linspace(-3, 3, 5)
X, Y, Z = np.meshgrid(x, y, z)

# Convert grid coordinates back to real meters for physics calculations
X_m, Y_m, Z_m = X/scale, Y/scale, Z/scale

# --- 3. Physics Calculations (Vectorized) ---
# Field from Plus Charge at (-2, 0, 0)
rx_plus = X_m - (-2e-10)
ry_plus = Y_m - 0
rz_plus = Z_m - 0
r_mag_plus = np.sqrt(rx_plus**2 + ry_plus**2 + rz_plus**2)
# Avoid division by zero at the exact charge location
r_mag_plus[r_mag_plus == 0] = np.nan

E_plus_x = (kcoulomb * qproton / r_mag_plus**3) * rx_plus
E_plus_y = (kcoulomb * qproton / r_mag_plus**3) * ry_plus
E_plus_z = (kcoulomb * qproton / r_mag_plus**3) * rz_plus

# Field from Minus Charge at (2, 0, 0)
rx_minus = X_m - 2e-10
ry_minus = Y_m - 0
rz_minus = Z_m - 0
r_mag_minus = np.sqrt(rx_minus**2 + ry_minus**2 + rz_minus**2)
r_mag_minus[r_mag_minus == 0] = np.nan

E_minus_x = (kcoulomb * -qproton / r_mag_minus**3) * rx_minus
E_minus_y = (kcoulomb * -qproton / r_mag_minus**3) * ry_minus
E_minus_z = (kcoulomb * -qproton / r_mag_minus**3) * rz_minus

# Total Field (Superposition)
Ex = E_plus_x + E_minus_x
Ey = E_plus_y + E_minus_y
Ez = E_plus_z + E_minus_z

# Normalize arrow sizes so they look neat on screen
E_total_mag = np.sqrt(Ex**2 + Ey**2 + Ez**2)
Ex, Ey, Ez = Ex/E_total_mag, Ey/E_total_mag, Ez/E_total_mag

# --- 4. Plotting ---
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(projection='3d')

# Plot the physical charges
ax.scatter(-2, 0, 0, color='red', s=200, label='Plus Charge')
ax.scatter(2, 0, 0, color='blue', s=200, label='Minus Charge')

# Draw ALL grid field arrows simultaneously with a single clean command
ax.quiver(X, Y, Z, Ex, Ey, Ez, length=0.5, color='orange', alpha=0.6, pivot='tail')

# Setup bounds
ax.set_xlim([-4, 4]); ax.set_ylim([-4, 4]); ax.set_zlim([-4, 4])
ax.set_xlabel('X ($10^{-10}$ m)'); ax.set_ylabel('Y ($10^{-10}$ m)'); ax.set_zlabel('Z ($10^{-10}$ m)')
ax.legend()

plt.show()
```
