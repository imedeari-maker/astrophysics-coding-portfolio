```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import griddata

# 1. Exact 18 experimental data points from Part C (X, Y, Voltage)
data = np.array([
# Circle Points (Negative Electrode around Y=10, Y=11)
[15, 10, 2.146], [15, 11, 2.929], [14, 11, 2.674], [14, 10, 2.636],
[13, 10, 1.751], [13, 9, 3.191], [14, 9, 2.655], [15, 9, 3.450],

# Bar Points (Positive Electrode along Y=3)
[23, 3, 11.01], [21, 3, 11.54], [19, 3, 11.45], [17, 3, 10.52],
[15, 3, 9.65], [13, 3, 9.14], [11, 3, 8.81], [9, 3, 9.07],
[7, 3, 9.77], [5, 3, 9.75]
])

x_measured, y_measured, v_measured = data[:, 0], data[:, 1], data[:, 2]

# 2. Recreate 26x20 lab grid
x_grid = np.linspace(0, 26, 200)
y_grid = np.linspace(0, 20, 200)
X, Y = np.meshgrid(x_grid, y_grid)

# 3. Fill the gaps between your measurements
V = griddata((x_measured, y_measured), v_measured, (X, Y), method='cubic')
V = np.clip(V, 1.5, 12.0) # Keep values strictly within your experimental range

# 4. Plot the 2D Equipotential Map
fig, ax = plt.subplots(figsize=(10, 8), dpi=300)
ax.set_facecolor('#ffffff')

# Draw physical electrodes so the plot matches lab setup paper
circle = plt.Circle((14.0, 10.0), 0.8, color='#2c3e50', zorder=5)
ax.add_patch(circle)
ax.plot([5, 23], [3.0, 3.0], color='#2c3e50', linewidth=10, solid_capstyle='butt', zorder=5)

# Draw the smooth contour lines at 1-Volt steps (from 2V to 11V)
levels = np.arange(2.0, 12.0, 1.0)
contours = ax.contour(X, Y, V, levels=levels, cmap='plasma', linewidths=1.5, zorder=3)

# Third Attempt
ax.clabel(contours, inline=True, fmt='%.1fV', colors='black')

# Grid and axis formatting to match your canvas template perfectly
ax.set_xlim(0, 26)
ax.set_ylim(0, 20)
ax.set_xticks(np.arange(0, 27, 2))
ax.set_yticks(np.arange(0, 21, 2))
ax.set_xlabel('X-Coordinate (cm)', fontsize=11, fontweight='bold')
ax.set_ylabel('Y-Coordinate (cm)', fontsize=11, fontweight='bold')
ax.set_title('2-D Equipotential Contour Map (Part C)\nDerived from Real Laboratory Measurements', fontsize=12, fontweight='bold', pad=15)
ax.grid(True, which='both', color='#e0e0e0', linestyle='--', linewidth=0.5)
ax.set_aspect('equal')

# Save the high-res 2D map image directly to folder
plt.savefig('2d_partC_contours.png', bbox_inches='tight', dpi=300)
plt.show()
```
